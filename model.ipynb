## Segmentation / Clustering (example)

```
from sklearn.cluster import KMeans
X = df[['loan_to_income','credit_score','utilization']].fillna(0)
km = KMeans(n_clusters=4, random_state=42).fit(X)
df['cluster'] = km.labels_
df.groupby('cluster')['default'].mean()
```
